```python
CREATE TABLE stars289_data (
    time TIMESTAMP,
    enclosure_temperature DOUBLE PRECISION,
    sky_temperature DOUBLE PRECISION,
    frequency DOUBLE PRECISION,
    msas DOUBLE PRECISION,
    zp DOUBLE PRECISION,
    sequence_number BIGINT,
    sun_alt DOUBLE PRECISION,
    moon_alt DOUBLE PRECISION,
    moon_illumination DOUBLE PRECISION
);
```

```python
import pandas as pd
from sqlalchemy import create_engine
import os

# Database connection settings
db_user = "postgres"
db_password = "1234"
db_host = "localhost"
db_port = "5433"
db_name = "TESS-DB"

# Create a connection to PostgreSQL
engine = create_engine(f'postgresql://{db_user}:{db_password}@{db_host}:{db_port}/{db_name}')

# Path to the folder containing ECSV files
ecsv_folder = 'jupyter/ECSV/stars289/'

# Loop through all files in the folder
for file_name in os.listdir(ecsv_folder):
    if file_name.endswith('.ecsv'):
        ecsv_file = os.path.join(ecsv_folder, file_name)

        # Read the ECSV file into a pandas DataFrame
        data = pd.read_csv(ecsv_file, comment='#', delimiter=',')

        # Rename columns to match the PostgreSQL table column names
        data.columns = [
            'time', 
            'enclosure_temperature', 
            'sky_temperature', 
            'frequency', 
            'msas', 
            'zp', 
            'sequence_number', 
            'sun_alt', 
            'moon_alt', 
            'moon_illumination'
        ]

        # Insert the data into the PostgreSQL table
        data.to_sql('stars289_data', engine, if_exists='append', index=False)
        print(f"Data from {file_name} has been successfully imported.")

print("All files have been imported into the PostgreSQL database.")
```

# Import data into SQL Lite DB

```python
import pandas as pd
from sqlalchemy import create_engine, text
import os

# Database connection settings for SQLite
db_name = 'stars289_data.db'

# Create an SQLite engine
engine = create_engine(f'sqlite:///{db_name}')

# Create the stars289_data table in SQLite
create_table_query = '''
CREATE TABLE IF NOT EXISTS stars289_data (
    time TIMESTAMP,
    enclosure_temperature DOUBLE PRECISION,
    sky_temperature DOUBLE PRECISION,
    frequency DOUBLE PRECISION,
    msas DOUBLE PRECISION,
    zp DOUBLE PRECISION,
    sequence_number BIGINT,
    sun_alt DOUBLE PRECISION,
    moon_alt DOUBLE PRECISION,
    moon_illumination DOUBLE PRECISION
);
'''

# Execute the table creation query
with engine.connect() as connection:
    connection.execute(text(create_table_query))  # Wrap the query in `text()`

# Path to the folder containing ECSV files
ecsv_folder = 'jupyter/ECSV/stars289/'

# Loop through all files in the folder and insert data into the SQLite table
for file_name in os.listdir(ecsv_folder):
    if file_name.endswith('.ecsv'):
        ecsv_file = os.path.join(ecsv_folder, file_name)

        # Read the ECSV file into a pandas DataFrame
        data = pd.read_csv(ecsv_file, comment='#', delimiter=',')

        # Rename columns to match the SQLite table column names
        data.columns = [
            'time', 
            'enclosure_temperature', 
            'sky_temperature', 
            'frequency', 
            'msas', 
            'zp', 
            'sequence_number', 
            'sun_alt', 
            'moon_alt', 
            'moon_illumination'
        ]

        # Insert the data into the SQLite table
        data.to_sql('stars289_data', engine, if_exists='append', index=False)
        print(f"Data from {file_name} has been successfully imported.")

print("All files have been imported into the SQLite database.")
```

## Check if data is missing in the time span

```python
import pandas as pd
from sqlalchemy import create_engine, text
from datetime import datetime, timedelta

# Database connection settings for SQLite
db_name = 'stars289_data.db'

# Create an SQLite engine
engine = create_engine(f'sqlite:///{db_name}')

# Define the time range (you can adjust these dates)
start_date = datetime(2024, 1, 1)  # Start date of the range
end_date = datetime(2024, 12, 31)  # End date of the range

# Generate a list of all days within the time range
all_days = pd.date_range(start=start_date, end=end_date, freq='D')

# Query the database for each day and check if at least one entry exists
missing_days = []

with engine.connect() as connection:
    for day in all_days:
        # Check if there is at least one entry for the current day
        query = text(f'''
        SELECT 1 FROM stars289_data
        WHERE DATE(time) = '{day.date()}'
        LIMIT 1;
        ''')
        result = connection.execute(query).fetchone()

        if result is None:
            missing_days.append(day.date())  # Add to missing days if no entry found

# Output the missing days
if missing_days:
    print("The following days are missing from the database:")
    for missing_day in missing_days:
        print(missing_day)
else:
    print("No missing days within the given time span.")
```
